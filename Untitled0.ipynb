```python
# ============================================================
# GOOGLE PAGERANK MINI (B2)
# Imports
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Optional: SymPy for checking the eigenvalue/eigenvector
import sympy as sp

# Display settings
np.set_printoptions(precision=6, suppress=True)
pd.set_option("display.precision", 6)

print("Libraries imported successfully!")
```

```python
# ============================================================
# USER CONFIGURATION
# ============================================================

# Five web pages
pages = ["Page A", "Page B", "Page C", "Page D", "Page E"]

# Links between pages
#
# Example:
# "Page A": ["Page B", "Page C"]
# means Page A contains links to Page B and Page C.
#
# IMPORTANT:
# A page does NOT automatically link to itself unless specified.

links = {
    "Page A": ["Page B", "Page C"],
    "Page B": ["Page C", "Page D"],
    "Page C": ["Page A"],
    "Page D": ["Page C", "Page E"],
    "Page E": ["Page C"]
}

# Power iteration configuration
tolerance = 1e-10
max_iterations = 1000

# Starting PageRank vector
# We start with equal probability for every page.
initial_rank = np.ones(len(pages)) / len(pages)

print("Pages:")
for page in pages:
    print(" ", page)

print("\nLinks:")
for page, destinations in links.items():
    print(f"  {page} -> {', '.join(destinations)}")

print("\nInitial rank:", initial_rank)
```

```python
# ============================================================
# INPUT VALIDATION
# ============================================================

def validate_inputs(pages, links):
    """Check that the configured web graph is valid."""

    if len(pages) == 0:
        raise ValueError("At least one page is required.")

    if len(set(pages)) != len(pages):
        raise ValueError("Page names must be unique.")

    for page in pages:
        if page not in links:
            raise ValueError(f"Missing link information for {page}")

        for destination in links[page]:
            if destination not in pages:
                raise ValueError(
                    f"{page} links to unknown page: {destination}"
                )

    print("Input validation: PASSED")


validate_inputs(pages, links)
```

```python
# ============================================================
# MATHEMATICAL MODEL
# Construct the column-stochastic transition matrix P
# ============================================================

def build_transition_matrix(pages, links):
    """
    Construct the PageRank transition matrix.

    P[i, j] = probability of moving from page j to page i.

    Therefore:
        P @ rank = next_rank
    """

    n = len(pages)
    page_index = {page: i for i, page in enumerate(pages)}

    P = np.zeros((n, n))

    for source_page in pages:

        source_index = page_index[source_page]
        destinations = links[source_page]

        # ----------------------------------------------------
        # Dangling-page assumption:
        # If a page has no outgoing links, distribute its
        # probability equally among all pages.
        # ----------------------------------------------------
        if len(destinations) == 0:

            P[:, source_index] = 1 / n

        else:

            probability = 1 / len(destinations)

            for destination in destinations:

                destination_index = page_index[destination]

                P[destination_index, source_index] = probability

    return P


P = build_transition_matrix(pages, links)

print("Transition Matrix P:\n")
print(P)

print("\nColumn sums:")
print(P.sum(axis=0))
```

```python
# ============================================================
# READABLE TRANSITION MATRIX
# ============================================================

transition_table = pd.DataFrame(
    P,
    index=pages,
    columns=pages
)

print("Transition Probability Matrix")
display(transition_table)

print("\nEvery column should sum to 1:")
display(
    pd.DataFrame(
        P.sum(axis=0),
        index=pages,
        columns=["Column Sum"]
    )
)
```

```python
# ============================================================
# POWER ITERATION
# ============================================================

def power_iteration(P, initial_rank, tolerance=1e-10, max_iterations=1000):
    """
    Find the steady-state PageRank vector using power iteration.

    Mathematical operation:
        r_(k+1) = P @ r_k

    Stop when:
        ||r_(k+1) - r_k||_1 < tolerance
    """

    rank = initial_rank.astype(float).copy()

    history = [rank.copy()]

    for iteration in range(1, max_iterations + 1):

        new_rank = P @ rank

        # Difference between consecutive vectors
        error = np.linalg.norm(new_rank - rank, ord=1)

        history.append(new_rank.copy())

        rank = new_rank

        if error < tolerance:
            return rank, iteration, error, np.array(history)

    return rank, max_iterations, error, np.array(history)


rank, iterations, final_error, history = power_iteration(
    P,
    initial_rank,
    tolerance,
    max_iterations
)

print("Power iteration completed!")
print(f"Iterations: {iterations}")
print(f"Final error: {final_error:.12e}")
```

```python
# ============================================================
# FINAL PAGERANK RESULTS
# ============================================================

result = pd.DataFrame({
    "Page": pages,
    "PageRank": rank,
    "Percentage": rank * 100
})

# Sort from most important to least important
result = result.sort_values(
    by="PageRank",
    ascending=False
).reset_index(drop=True)

print("FINAL PAGERANK")
display(result)

most_important = result.iloc[0]["Page"]

print(f"\nMost important page: {most_important}")
print(f"PageRank = {result.iloc[0]['PageRank']:.6f}")
print(f"Percentage = {result.iloc[0]['Percentage']:.2f}%")
```

```python
# ============================================================
# EIGENVALUE = 1 VERIFICATION
# ============================================================

steady_state = P @ rank

eigen_error = np.linalg.norm(
    steady_state - rank,
    ord=1
)

print("Steady-state verification")
print("-" * 40)

print("PageRank vector r:")
print(rank)

print("\nP @ r:")
print(steady_state)

print("\nDifference ||P r - r||₁:")
print(f"{eigen_error:.12e}")

if eigen_error < tolerance:
    print("\n✓ P r ≈ r")
    print("✓ Therefore P r = 1 r approximately")
    print("✓ The PageRank vector is the eigenvector for λ = 1")
else:
    print("\nThe iteration has not reached the required tolerance.")
```

```python
# ============================================================
# INDEPENDENT EIGENVALUE CHECK
# ============================================================

eigenvalues, eigenvectors = np.linalg.eig(P)

print("Eigenvalues of P:\n")

for i, value in enumerate(eigenvalues):
    print(f"λ{i+1} = {value}")

# Find eigenvalue closest to 1
closest_index = np.argmin(np.abs(eigenvalues - 1))

closest_eigenvalue = eigenvalues[closest_index]
eigenvector = np.real(eigenvectors[:, closest_index])

# Eigenvectors can have arbitrary scale/sign.
# Normalize it so that the components sum to 1.
eigenvector = eigenvector / eigenvector.sum()

print("\nEigenvalue closest to 1:")
print(closest_eigenvalue)

print("\nNormalized eigenvector for λ ≈ 1:")
print(eigenvector)

print("\nPower iteration result:")
print(rank)
```

```python
# ============================================================
# POWER ITERATION CONVERGENCE TABLE
# ============================================================

convergence_data = []

for i, vector in enumerate(history):
    convergence_data.append({
        "Iteration": i,
        **{
            pages[j]: vector[j]
            for j in range(len(pages))
        },
        "Sum": vector.sum()
    })

convergence_table = pd.DataFrame(convergence_data)

print("First iterations:")
display(convergence_table.head(10))

print("\nLast iterations:")
display(convergence_table.tail(10))
```

```python
# ============================================================
# BENCHMARK TEST
# ============================================================

benchmark_pages = ["A", "B", "C", "D", "E"]

benchmark_links = {
    "A": ["B", "C"],
    "B": ["C", "D"],
    "C": ["A"],
    "D": ["C", "E"],
    "E": ["C"]
}

benchmark_P = build_transition_matrix(
    benchmark_pages,
    benchmark_links
)

benchmark_initial = (
    np.ones(len(benchmark_pages))
    / len(benchmark_pages)
)

benchmark_rank, benchmark_iterations, benchmark_error, _ = (
    power_iteration(
        benchmark_P,
        benchmark_initial,
        tolerance=1e-10,
        max_iterations=1000
    )
)

benchmark_result = pd.DataFrame({
    "Page": benchmark_pages,
    "PageRank": benchmark_rank
})

print("Benchmark dataset:")
print("A → B, C")
print("B → C, D")
print("C → A")
print("D → C, E")
print("E → C")

print("\nBenchmark PageRank:")
display(benchmark_result)

print(f"Iterations: {benchmark_iterations}")
print(f"Final error: {benchmark_error:.3e}")

print(
    "\nBenchmark most important page:",
    benchmark_pages[np.argmax(benchmark_rank)]
)
```

```python
# ============================================================
# ANTI-BUG / SAFETY CHECK CELL
# ============================================================

def anti_bug_check(pages, links, P, rank, tolerance=1e-10):

    print("Running PageRank anti-bug checks...\n")

    # 1. Check number of pages
    assert len(pages) > 0, "ERROR: No pages found."

    # 2. Check duplicate page names
    assert len(pages) == len(set(pages)), \
        "ERROR: Duplicate page names detected."

    # 3. Check every page has link information
    for page in pages:
        assert page in links, \
            f"ERROR: Missing links for {page}"

    # 4. Check all destinations are valid
    for source in pages:
        for destination in links[source]:
            assert destination in pages, \
                f"ERROR: {source} links to unknown page {destination}"

    # 5. Check matrix dimensions
    n = len(pages)

    assert P.shape == (n, n), \
        "ERROR: Transition matrix has incorrect dimensions."

    # 6. Check matrix contains valid numbers
    assert np.all(np.isfinite(P)), \
        "ERROR: Matrix contains NaN or infinite values."

    # 7. Check transition probabilities are non-negative
    assert np.all(P >= 0), \
        "ERROR: Negative transition probability detected."

    # 8. Check columns sum to 1
    column_sums = P.sum(axis=0)

    assert np.allclose(column_sums, 1.0, atol=tolerance), \
        "ERROR: Transition matrix columns do not sum to 1."

    # 9. Check PageRank vector
    assert len(rank) == n, \
        "ERROR: PageRank vector has incorrect length."

    # 10. Check PageRank values are valid
    assert np.all(np.isfinite(rank)), \
        "ERROR: PageRank contains NaN or infinite values."

    # 11. PageRank cannot be negative
    assert np.all(rank >= -tolerance), \
        "ERROR: Negative PageRank value detected."

    # 12. PageRank probabilities must sum to 1
    assert np.isclose(rank.sum(), 1.0, atol=tolerance), \
        "ERROR: PageRank values do not sum to 1."

    # 13. Check eigenvector condition P*r = r
    eigenvector_error = np.linalg.norm(P @ rank - rank, ord=1)

    assert eigenvector_error < tolerance, \
        "ERROR: PageRank has not reached the steady state."

    print("✓ All anti-bug checks PASSED!")
    print("✓ Input data is valid")
    print("✓ Transition matrix is valid")
    print("✓ Probabilities are valid")
    print("✓ PageRank sums to 1")
    print("✓ P @ rank ≈ rank")
    print("✓ Eigenvalue λ = 1 condition verified")
    print(f"\nMaximum verification error: {eigenvector_error:.3e}")


# Run anti-bug checks
anti_bug_check(
    pages,
    links,
    P,
    rank,
    tolerance
)
```
